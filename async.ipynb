```python
import time

def ambil_data_a():
    time.sleep(3)  # nunggu 3 detik, program berhenti total
    return "data A"

def ambil_data_b():
    time.sleep(4)
    return "data B"

a = ambil_data_a()  # nunggu 3 detik
b = ambil_data_b()  # nunggu 4 detik lagi
# total: 7 detik
```

```python
import asyncio

async def ambil_data_a():
    await asyncio.sleep(2)  # "istirahat", biarkan tugas lain jalan
    return "data A"

async def ambil_data_b():
    await asyncio.sleep(2)
    return "data B"

a, b = await asyncio.gather(ambil_data_a(), ambil_data_b())
# total: ~2 detik, karena keduanya "menunggu" bersamaan
```

```python
import asyncio
import time

async def rebus_air():
    print("Mulai rebus air...")
    await asyncio.sleep(4)
    print("Air mendidih!")

async def goreng_telur():
    print("Mulai goreng telur...")
    await asyncio.sleep(2)
    print("Telur matang!")

async def main():
    mulai = time.time()
    await rebus_air()      # tunggu 4 detik dulu, baru lanjut
    await goreng_telur()   # baru mulai setelah rebus air selesai
    print(f"Total waktu: {time.time() - mulai:.1f} detik")

await main()
```

```python
async def main():
    mulai = time.time()
    await asyncio.gather(
        rebus_air(),
        goreng_telur(),
    )
    print(f"Total waktu: {time.time() - mulai:.1f} detik")

await main()
```

```python
import asyncio

async def run_1():
    print(f"run 1 dimulai")
    await asyncio.sleep(5)
    print(f"============= run 1 selesai =============")
    print(f"run 1 selesai")

async def run_2():
    print(f"run 2 dimulai")
    await asyncio.sleep(3)
    print(f"============= run 2 selesai =============")
    print(f"run 2 selesai")

async def main():
    await asyncio.gather(
        run_1(),
        run_2(),
    )

await main()
```
